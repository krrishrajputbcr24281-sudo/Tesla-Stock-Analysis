```python
import seaborn as sns
```

```python
missing_values = df.isnull().sum()
```

```python
missing_df = missing_values.reset_index()
missing_df.columns = ['Attribute', 'Missing Values']
```

```python
plt.figure(figsize=(10, 6))
sns.barplot(
    data=missing_df,
    x='Attribute',
    y='Missing Values',
    color='steelblue'
)

plt.title('Number of Missing Values by Attribute')
plt.xlabel('Attributes')
plt.ylabel('Number of Missing Values')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
df = pd.read_csv("C:/Users/admin/Downloads/TSLA.csv")

result = df.loc[df["High"] > 450, ["Date", "Open"]].reset_index(drop=True)


print(result)
```

```python
top10 = df.nlargest(10, "Close")

result = df.nlargest(10, "Close")[["Date", "High"]].reset_index(drop=True)


print(result)
```

```python
bottom10 = df.nsmallest(10, "Open")

result = bottom10[["Date", "Low"]].reset_index(drop=True)

print(result)
```

```python
sns.lineplot(x="Date", y="High", data=top10)

plt.xlabel("Date")
plt.ylabel("High")
plt.title("Date versus High")
plt.xticks(rotation=45)
plt.show()
```

```python
sns.lineplot(x="Date", y="High", data=bottom10)

plt.xlabel("Date")
plt.ylabel("High")
plt.title("Date versus High")
plt.xticks(rotation=45)
plt.show()
```

```python
result = df.loc[(df["Volume"] >= 1000000) & (df["Volume"] <= 1500000),
                ["Date", "Volume"]].reset_index(drop=True)
print(result)
```

```python
result['Date'] = pd.to_datetime(result['Date'])
```

```python
plt.figure(figsize=(12, 5))

sns.barplot(x='Date', y='Volume', data=result,palette=sns.color_palette("viridis"))

plt.xlabel('Date')
plt.ylabel('Volume')
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
result = df.loc[df["Open"]==df["Close"],["Date","High","Low"]].reset_index(drop=True)
print(result)
```

```python
columns = ["Open", "High", "Low", "Close", "Volume"]

fig, axes = plt.subplots(3, 2, figsize=(12, 10))
axes = axes.flatten()

for i, column in enumerate(columns):
    axes[i].hist(df[column].dropna(), bins=30,
                 color="steelblue", edgecolor="black")
    axes[i].set_title("Distribution of " + column)
    axes[i].set_xlabel(column)
    axes[i].set_ylabel("Count")
axes[5].set_visible(False)

plt.tight_layout()
plt.show()
```

```python
sns.set_style("whitegrid")


fig, axes = plt.subplots(1, 3, figsize=(18, 5))

sns.scatterplot(data=df, x="Open", y="Close", ax=axes[0])
axes[0].set_title("Open vs Close")
axes[0].set_xlabel("Open")
axes[0].set_ylabel("Close")


sns.scatterplot(data=df, x="High", y="Low", ax=axes[1])
axes[1].set_title("High vs Low")
axes[1].set_xlabel("High")
axes[1].set_ylabel("Low")

sns.scatterplot(data=df, x="Volume", y="Adj Close", ax=axes[2])
axes[2].set_title("Volume vs Adj Close")
axes[2].set_xlabel("Volume")
axes[2].set_ylabel("Adj Close")

plt.tight_layout()
plt.show()
```
